```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('canada_per_capita_income.csv')
df.head()
```

```python
plot = df.plot(x='year', y='income', kind='scatter')
plt.show()
```

```python
x = df[['year']]
y = df[['income']]
```

```python
reg = linear_model.LinearRegression()
reg.fit(x, y)
```

```python
reg.predict([[1990]])
```

```python
reg.score(x, y)
```
